```python
from pathlib import Path
from io import StringIO
import pandas as pd
import matplotlib.pyplot as plt
import os
import seaborn as sns
import matplotlib.cm as cm
import matplotlib.colors as mcolors

files = ["PLA_ECO_VERS_1x5_2x5_3x5.csv", "PLA_ECO_VERS_4x5_5x5_6x5.csv", "PLA_ECO_VERS_7x5_8x5_9x5.csv"]  
max_file_size = 19

all_dfs = []

for file in files:
    
    with open(file, "r") as f:
        lines = f.readlines()
        
    series_blocks = []
    current_block = []

    for line in lines:
        if line.strip().startswith(tuple(str(i) + ",Time" for i in range(1, max_file_size+1))):
            # save current block (if any)
            if current_block:
                series_blocks.append(current_block)
                current_block = []
        current_block.append(line)
        
    
    # add the last one
    if current_block:
        series_blocks.append(current_block)

    time_series_list = []
    max_length = []
    max_force = []
    
    file_size = len(series_blocks) - 1

    for block in series_blocks[1:]:
        # Find the series ID (first number before comma)
        header_line = block[0].strip()
        series_id = header_line.split(',')[0]

        # Skip the first 2 header lines (like "1,Time..." and ",(s),(mm),(kN)")
        data = ''.join(block[2:])

        # Use io.StringIO instead of pandas.compat.StringIO
        df = pd.read_csv(StringIO(data), header=None, names=["_", "Time", "Displacement", "Force"])
        df = df.drop(columns=["_"])  # drop empty index column
        df["Series"] = series_id
        time_series_list.append(df)
        max_length.append(df["Displacement"].max())
        max_force.append(df["Force"].max())
        
    start = None
    for i, line in enumerate(lines):
        if line.strip() == "Results Table 1":
            start = i + 3
            break
            
    table_lines = lines[(start) : (start + file_size)]
    header = lines[start-2].split(",")
    df = pd.read_csv(StringIO("".join(table_lines)), names = header)
    df["Elongation at break (%)"] = [elon * 100/60 for elon in max_length]
    df["Maximum force (kN)"] = [force for force in max_force]
    all_dfs.append(df)
        

final_df = pd.concat(all_dfs, ignore_index=True)

cols = [
    "Tensile stress at Yield (Offset 0.2 %)",
    "Modulus (Automatic Young's)",
    "Elongation at break (%)",
    "Maximum force (kN)"
]

final_df[cols] = final_df[cols].apply(pd.to_numeric, errors="coerce")

material = pd.read_csv("PLA_ECO_VERS_guide.csv")
final_df = pd.concat([final_df, material], axis = 1)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(20, 10))

lst = final_df.Material.to_list()

order = list(dict.fromkeys(lst))

final_df['Material'] = pd.Categorical(final_df['Material'], categories=order, ordered=True)
final_df_sorted = final_df.sort_values('Material')

color_property = "PLA"

norm = mcolors.Normalize(vmin=final_df[color_property].min(), 
                         vmax=final_df[color_property].max())
cmap = cm.coolwarm


sns.boxplot(
    data=final_df,
    x=final_df.Material,
    y="Tensile stress at Yield (Offset 0.2 %)",
    ax=axes[0,0],
    hue=color_property,
    width=0.8,
    dodge=False,
    palette = "coolwarm",
    legend = False
)


# 3. Create a ScalarMappable and attach the colorbar
sm = cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])  # required dummy call
fig.colorbar(sm, ax=axes[0, 0], label=color_property)

axes[0,0].set_title("Tensile Stress at Yield")
axes[0,0].set_xlabel("Material")
axes[0,0].set_ylabel("Tensile Stress at Yield (MPa)")

sns.boxplot(
    data=final_df,
    x=final_df.Material,
    y="Modulus (Automatic Young\'s)",
    ax=axes[0,1],
    hue=color_property,
    width=0.8,
    dodge=False,
    palette = "coolwarm",
    
)

fig.colorbar(sm, ax=axes[0, 1], label=color_property)

axes[0,1].set_title("Modulus (Automatic Young\'s)")
axes[0,1].set_xlabel("Material")
axes[0,1].set_ylabel("Modulus (MPa)")
plt.suptitle("")

sns.boxplot(
    data=final_df,
    x=final_df.Material,
    y="Elongation at break (%)",
    ax=axes[1,0],
    hue=color_property,
    width=0.8,
    dodge=False,
    palette = "coolwarm"
)

fig.colorbar(sm, ax=axes[1, 0], label=color_property)

axes[1,0].set_title("Elongation at break (%)")
axes[1,0].set_xlabel("Material")
axes[1,0].set_ylabel("Elongation at break (%)")
plt.suptitle("")

sns.boxplot(
    data=final_df,
    x=final_df.Material,
    y="Maximum force (kN)",
    ax=axes[1,1],
    hue=color_property,
    width=0.8,
    dodge=False,
    palette = "coolwarm"
)
fig.colorbar(sm, ax=axes[1,1], label=color_property)

axes[1,1].set_title("Maximum force (kN)")
axes[1,1].set_xlabel("Material")
axes[1,1].set_ylabel("Maximum force (kN)")
plt.suptitle("")
plt.show()
```

```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from botorch.models import SingleTaskGP
from botorch.fit import fit_gpytorch_mll
from botorch.acquisition import ExpectedImprovement
from botorch.optim import optimize_acqf
from gpytorch.mlls import ExactMarginalLogLikelihood

for property in properties:
    # ── 1. Load data from DataFrame ───────────────────────────────────────────────
    
    # BoTorch expects (n, d) and (n, 1), dtype float64 → float32 is fine too
    train_X = torch.tensor(final_df[["Ecoflex", "Versaloid"]].values, dtype=torch.double)
    train_Y = torch.tensor(final_df[[property]].values, dtype=torch.double)
    
    # Standardize Y (recommended for GP)
    Y_mean, Y_std = train_Y.mean(), train_Y.std()
    train_Y_norm = (train_Y - Y_mean) / Y_std
    
    # ── 2. Fit GP model ───────────────────────────────────────────────────────────
    model = SingleTaskGP(train_X, train_Y_norm)
    mll = ExactMarginalLogLikelihood(model.likelihood, model)
    fit_gpytorch_mll(mll)
    model.eval()
    
    # ── 3. Define acquisition function ───────────────────────────────────────────
    best_f = train_Y_norm.max()
    EI = ExpectedImprovement(model=model, best_f=best_f, maximize=True)
    
    # ── 4. Optimize acquisition → next candidate ─────────────────────────────────
    bounds = torch.tensor([[0.0, 0.0], [0.3, 0.3]], dtype=torch.double)
    candidate, acq_value = optimize_acqf(
        acq_function=EI,
        bounds=bounds,
        q=1,
        num_restarts=10,
        raw_samples=64,
    )
    print(f"Next candidate: x1={candidate[0,0]:.4f}, x2={candidate[0,1]:.4f}")
    print(f"EI value: {acq_value.item():.4f}")
    
    # ── 5. Plot posterior mean, variance, EI map ─────────────────────────────────
    res = 80  # grid resolution
    x1g = torch.linspace(0, 0.3, res, dtype=torch.double)
    x2g = torch.linspace(0, 0.3, res, dtype=torch.double)
    X1, X2 = torch.meshgrid(x1g, x2g, indexing="ij")
    grid = torch.stack([X1.flatten(), X2.flatten()], dim=-1)  # (res², 2)
    
    with torch.no_grad():
        posterior = model.posterior(grid)
        mean = posterior.mean.reshape(res, res).numpy()
        var  = posterior.variance.reshape(res, res).numpy()
    
    # EI on grid
    with torch.no_grad():
        ei_vals = EI(grid.unsqueeze(1)).reshape(res, res).numpy()
    
    # Back-transform mean to original scale for readability
    mean_orig = mean * Y_std.item() + Y_mean.item()
    
    x1_np = x1g.numpy()
    x2_np = x2g.numpy()
    obs_x1 = final_df["Ecoflex"].values
    obs_x2 = final_df["Versaloid"].values
    
    fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
    cbar_kw = dict(fraction=0.046, pad=0.04)
    
    # --- Posterior mean ---
    ax = axes[0]
    c = ax.contourf(x1_np, x2_np, mean_orig.T, levels=30, cmap="magma")
    fig.colorbar(c, ax=ax, **cbar_kw)
    ax.scatter(obs_x1, obs_x2, c="white", edgecolors="black", s=60, zorder=5, label="observations")
    ax.set_title("Posterior Mean", fontsize=13)
    ax.set_xlabel("Ecoflex"); ax.set_ylabel("Versaloid")
    ax.legend(fontsize=8)
    
    # --- Posterior variance ---
    ax = axes[1]
    c = ax.contourf(x1_np, x2_np, var.T, levels=30, cmap="magma")
    fig.colorbar(c, ax=ax, **cbar_kw)
    ax.scatter(obs_x1, obs_x2, c="white", edgecolors="black", s=60, zorder=5)
    ax.set_title("Posterior Variance", fontsize=13)
    ax.set_xlabel("Ecoflex"); ax.set_ylabel("Versaloid")
    
    # --- Expected Improvement ---
    ax = axes[2]
    c = ax.contourf(x1_np, x2_np, ei_vals.T, levels=30, cmap="magma")
    fig.colorbar(c, ax=ax, **cbar_kw)
    ax.scatter(obs_x1, obs_x2, c="white", edgecolors="black", s=60, zorder=5, label="observations")
    ax.scatter(*candidate[0].numpy(), marker="*", c="red", s=250, zorder=6, label="next candidate")
    ax.set_title("Expected Improvement", fontsize=13)
    ax.set_xlabel("Ecoflex"); ax.set_ylabel("Versaloid")
    ax.legend(fontsize=8)
    
    plt.suptitle(property, fontsize=14, fontweight="bold")
    plt.tight_layout()
    plt.savefig("bo_maps.png", dpi=150, bbox_inches="tight")
    plt.show()
```

```python
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from botorch.models import SingleTaskGP
from botorch.models.model_list_gp_regression import ModelListGP
from botorch.fit import fit_gpytorch_mll
from botorch.acquisition.multi_objective.logei import qLogNoisyExpectedHypervolumeImprovement as qLogNEHVI
from botorch.optim import optimize_acqf
from gpytorch.mlls.sum_marginal_log_likelihood import SumMarginalLogLikelihood
from botorch.models.transforms.input import Normalize

MO_properties = ['Tensile stress at Yield (Offset 0.2 %)', 'Elongation at break (%)']


# ── 1. Aggregate repeated compositions (mean per unique X) ────────────────────
# This fixes the near-singular kernel matrix from duplicate X rows
agg_df = final_df.groupby(["Ecoflex", "Versaloid"])[MO_properties].mean().reset_index()

train_X = torch.tensor(agg_df[["Ecoflex", "Versaloid"]].values, dtype=torch.double)
train_Y = torch.tensor(agg_df[MO_properties].values, dtype=torch.double)

# ── 2. Normalize outputs ──────────────────────────────────────────────────────
Y_mean = train_Y.mean(dim=0)
Y_std  = train_Y.std(dim=0).clamp(min=1e-6)   # avoid div/0
train_Y_norm = (train_Y - Y_mean) / Y_std

# ── 3. Fit one GP per output ──────────────────────────────────────────────────
models = [
    SingleTaskGP(
        train_X, train_Y_norm[:, i : i + 1],
        input_transform=Normalize(d=train_X.shape[-1]),
    )
    for i in range(len(MO_properties))
]
model = ModelListGP(*models)
mll   = SumMarginalLogLikelihood(model.likelihood, model)
fit_gpytorch_mll(mll)
model.eval()

# ── 4. qLogNEHVI — suggest 5 points ──────────────────────────────────────────
Y_min = train_Y_norm.min(dim=0).values
Y_max = train_Y_norm.max(dim=0).values
ref_point = (Y_min - 0.1 * (Y_max - Y_min)).tolist()

acq = qLogNEHVI(
    model=model,
    ref_point=ref_point,
    X_baseline=train_X,
    prune_baseline=True,
)

# Bounds match your actual data range
x1_min, x1_max = float(0.0), float(0.3)
x2_min, x2_max = float(0.0), float(0.3)
bounds = torch.tensor([[x1_min, x2_min], [x1_max, x2_max]], dtype=torch.double)

print("Optimizing acquisition (this takes ~1-2 min)...")
candidates, acq_value = optimize_acqf(
    acq_function=acq,
    bounds=bounds,
    q=5,
    num_restarts=10,       # reduced from 10
    raw_samples=128,      # reduced but sufficient
    options={"maxiter": 200},
)

print("\n=== 5 suggested candidates ===")
for i, c in enumerate(candidates):
    print(f"  Point {i+1}: Ecoflex={c[0]:.4f}, Versaloid={c[1]:.4f}")

# ── 5. Posterior maps ─────────────────────────────────────────────────────────
res = 60  # lower than 80 — fast and still smooth
x1g = torch.linspace(x1_min, x1_max, res, dtype=torch.double)
x2g = torch.linspace(x2_min, x2_max, res, dtype=torch.double)
X1, X2 = torch.meshgrid(x1g, x2g, indexing="ij")
grid = torch.stack([X1.flatten(), X2.flatten()], dim=-1)

with torch.no_grad():
    posterior  = model.posterior(grid)
    mean       = posterior.mean.reshape(res, res, len(MO_properties)).numpy()
    var        = posterior.variance.reshape(res, res, len(MO_properties)).numpy()

mean_orig = mean * Y_std.numpy() + Y_mean.numpy()

obs_x1  = agg_df["Ecoflex"].values
obs_x2  = agg_df["Versaloid"].values
cand_np = candidates.detach().numpy()
x1_np   = x1g.numpy()
x2_np   = x2g.numpy()

fig, axes = plt.subplots(len(MO_properties), 2, figsize=(12, 5 * len(MO_properties)))

for i, name in enumerate(MO_properties):
    # Mean
    ax = axes[i, 0]
    c  = ax.contourf(x1_np, x2_np, mean_orig[:, :, i].T, levels=30, cmap="magma")
    fig.colorbar(c, ax=ax, fraction=0.046, pad=0.04)
    ax.scatter(obs_x1, obs_x2, c="white", edgecolors="black", s=60, zorder=5, label="obs")
    ax.scatter(cand_np[:, 0], cand_np[:, 1], marker="*", c="red", s=250, zorder=6, label="candidates")
    ax.set_title(f"{name} — Posterior Mean", fontsize=11)
    ax.set_xlabel("Ecoflex"); ax.set_ylabel("Versaloid")
    if i == 0:
        ax.legend(fontsize=8)

    # Variance
    ax = axes[i, 1]
    c  = ax.contourf(x1_np, x2_np, var[:, :, i].T, levels=30, cmap="magma")
    fig.colorbar(c, ax=ax, fraction=0.046, pad=0.04)
    ax.scatter(obs_x1, obs_x2, c="white", edgecolors="black", s=60, zorder=5)
    ax.scatter(cand_np[:, 0], cand_np[:, 1], marker="*", c="red", s=250, zorder=6)
    ax.set_title(f"{name} — Posterior Variance", fontsize=11)
    ax.set_xlabel("Ecoflex"); ax.set_ylabel("Versaloid")

plt.suptitle("Multi-Objective BO (qLogNEHVI) — 4 outputs, 5 candidates",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.savefig("bo_multi_output_maps.png", dpi=150, bbox_inches="tight")
plt.show()
```
